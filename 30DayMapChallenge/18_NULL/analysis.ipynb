```python
# pip install pygbif pandas
from pygbif import occurrences as occ
import pandas as pd

res = occ.search(issue='ZERO_COORDINATE', limit=300)
print(f"Total zero-coordinate records in GBIF: {res['count']:,}")

df = pd.DataFrame(res['results'])
cols = ['scientificName', 'kingdom', 'class', 'decimalLatitude',
        'decimalLongitude', 'country', 'basisOfRecord', 'institutionCode', 'year']
df = df[[c for c in cols if c in df.columns]]

print(df.head(20).to_string(index=False))
print("\nBy kingdom:\n", df['kingdom'].value_counts())
print("\nBy class:\n", df['class'].value_counts().head(10))
print("\nClaimed country:\n", df['country'].value_counts().head(10))

df.to_csv('null_island_sample.csv', index=False)
```
